```python
# (a) Count frequency of characters in a string
def char_freq(string):
    freq = {}  # empty dictionary
    for ch in string:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq

# (b) Count frequency of characters from a file
def char_freq_file(filename):
    freq = {}
    with open(filename, "r") as f:
        text = f.read()
        for ch in text:
            if ch in freq:
                freq[ch] += 1
            else:
                freq[ch] = 1
    return freq

# (c) Print histogram from frequency dictionary
def histogram(dict_freq):
    for ch, count in dict_freq.items():
        print(ch, ":", "*" * count)


# --- Main program with menu ---
if __name__ == "__main__":
    while True:
        print("\n--- This is a Character Frequency Program ---")
        print("(a) Count frequency from a string")
        print("(b) Count frequency from a file")
        print("(c) Exit")

        choice = input("Choose an option: ").lower()

        if choice == "a":
            user_word = input("Enter a word (e.g., sample): ")
            freq_user = char_freq(user_word)
            print("Frequencies:", freq_user)
            print("Histogram:")
            histogram(freq_user)

        elif choice == "b":
            filename = input("Enter the filename (e.g., sample.txt): ")
            try:
                freq2 = char_freq_file(filename)
                print("Frequencies:", freq2)
                print("Histogram:")
                histogram(freq2)
            except FileNotFoundError:
                print("File not found! Make sure the file is in the same folder.")

        elif choice == "c":
            print("Program terminated. Please try again!")
            break  # exit the loop

        else:
            print("Invalid choice! Please select a, b, or c to continue.")
```
